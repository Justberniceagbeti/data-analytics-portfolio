```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Date": [
        "2026-01-05","2026-01-10","2026-01-15",
        "2026-02-03","2026-02-11","2026-02-18",
        "2026-03-02","2026-03-10","2026-03-25"
    ],

    "Product": [
        "Laptop","Phone","Laptop",
        "Tablet","Phone","Laptop",
        "Phone","Tablet","Laptop"
    ],

    "Region": [
        "Lagos","Abuja","Lagos",
        "Port Harcourt","Abuja","Lagos",
        "Lagos","Port Harcourt","Abuja"
    ],

    "Revenue": [
        450000,220000,470000,
        180000,240000,500000,
        260000,200000,520000
    ]
}

df = pd.DataFrame(data)
df
```

```python
# Total Revenue
total_revenue = df["Revenue"].sum()

# Best Selling Product (by revenue)
best_product = df.groupby("Product")["Revenue"].sum().idxmax()

# Top Region
top_region = df.groupby("Region")["Revenue"].sum().idxmax()

print("📊 SALES PERFORMANCE SUMMARY")
print("----------------------------")
print("Total Revenue: ₦{:,.0f}".format(total_revenue))
print("Best Selling Product:", best_product)
print("Top Region:", top_region)
```

```python
product_sales = df.groupby("Product")["Revenue"].sum()

plt.figure(figsize=(8,4))
plt.bar(product_sales.index, product_sales.values)

plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₦)")

plt.show()
```

```python
region_sales = df.groupby("Region")["Revenue"].sum()

plt.figure(figsize=(8,4))
plt.bar(region_sales.index, region_sales.values)

plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (₦)")

plt.xticks(rotation=20)

plt.show()
```

```python
# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Create Month column
df["Month"] = df["Date"].dt.strftime("%b")

# Group revenue by month
monthly_sales = df.groupby("Month")["Revenue"].sum()

plt.figure(figsize=(8,4))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o", linewidth=2)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₦)")

plt.show()
```

```python
best_month = monthly_sales.idxmax()
best_month_value = monthly_sales.max()

lowest_month = monthly_sales.idxmin()
lowest_month_value = monthly_sales.min()

print("📌 BUSINESS INSIGHTS")
print("-------------------")
print(f"Best Sales Month: {best_month} (₦{best_month_value:,.0f})")
print(f"Lowest Sales Month: {lowest_month} (₦{lowest_month_value:,.0f})")
print(f"Total Revenue: ₦{df['Revenue'].sum():,.0f}")
print(f"Average Sale Per Transaction: ₦{df['Revenue'].mean():,.0f}")
```

```python
df.to_csv("clean_sales_data.csv", index=False)
print("Dataset saved successfully.")
```
