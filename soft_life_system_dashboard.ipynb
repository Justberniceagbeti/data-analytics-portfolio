```python
mood = "Tired"
energy = 3

def soft_plan(mood, energy):
    if energy <= 3:
        return [
            "Focus on 1 important task",
            "Take 2 short breaks",
            "Hydrate well",
            "Rest without guilt"
        ]
    elif energy <= 7:
        return [
            "Complete top 3 tasks",
            "Do light movement",
            "Journal tonight"
        ]
    else:
        return [
            "Do deep work",
            "Exercise",
            "Plan tomorrow"
        ]

plan = soft_plan(mood, energy)

for item in plan:
    print("-", item)
```

```python
import pandas as pd
import os
from datetime import datetime

mood = input("How do you feel today? ")
energy = int(input("Rate your energy from 1 to 10: "))

entry = pd.DataFrame([{
    "Date": datetime.now().strftime("%Y-%m-%d"),
    "Mood": mood,
    "Energy": energy
}])

file_name = "soft_life_data.csv"

if os.path.exists(file_name):
    entry.to_csv(file_name, mode="a", header=False, index=False)
else:
    entry.to_csv(file_name, mode="w", header=True, index=False)

print("✅ Entry saved successfully.")
```

```python
data = pd.read_csv("soft_life_data.csv")
data
```

```python
import matplotlib.pyplot as plt

# Make row count start from 1 for display
x = range(1, len(data) + 1)

plt.figure(figsize=(9,4))
plt.plot(x, data["Energy"], marker="o", linewidth=2)

plt.title("Energy Trend Across Mood Entries")
plt.xlabel("Entry Number")
plt.ylabel("Energy Level")

plt.xticks(x)
plt.ylim(0,10)

plt.show()
```

```python
print("Average Energy:", data["Energy"].mean())
print()
print("Mood Counts:")
print(data["Mood"].value_counts())
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,4))

mood_counts = data["Mood"].value_counts()

ax.bar(mood_counts.index, mood_counts.values)

ax.set_title("Mood Distribution")
ax.set_xlabel("Mood")
ax.set_ylabel("Count")

plt.xticks(rotation=30)
plt.show()
```
